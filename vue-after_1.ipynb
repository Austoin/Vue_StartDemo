### 模板引用

``` javascript
<script>
export default {
    data() {
        return {
           cnt: "content",

            contents: "HI JOJO",

        }
    },

    methods: {
       getDemo(){
        console.log(this.$refs.cont.innerHTML = "Hello JOJO")
        console.log(this.$refs.username.value)
       }

    }
}

</script>

<template>
    <div :class="cnt">
        <h2>模板引用</h2>

        <div ref="cont" class="cont">
            {{ contents }}
        </div>
        

        <input type="text" ref="username">

        <button @click="getDemo">获取元素</button>
        
    </div>


</template>

<style>

    .content {
        display: flex;
        flex-direction: column;
        align-items: center;   
        justify-content: center;
        
        height: 50vh; 
        border: 1px solid #000;
    }

</style>

### 组件传递数据

> 需要注意的是 **只能父传子不能子传父**

> 先是父文件

``` javascript
<script>
import Child from "./index_2_Child.vue"

export default {
    data() {
        return {
            cnt:"content",

            test:"动态数据",
        }
    },

    components:{
        Child
    }
}

</script>

<template>

    <div :class="cnt">
        <h2>组件传递数据-Parent</h2>

        <child title="Parent 数据传递" demo="测试" :jojo = "test" />
        
    </div>


</template>

<style scoped>
    .content {
        display: flex;
        flex-direction: column;
        align-items: center;   
        justify-content: center;
        
        height: 50vh; 
        border: 1px solid #000;
    }
</style>

> 然后是子程序

``` javascript
<script>
export default {
    data() {
        return {
            cnt:"rd",
        }
    },

    props:["title","demo","jojo"],
}

</script>

<template>
    <div>
        <h2>组件传递数据-Child</h2>

        <p>{{ title }}</p>
        <p>{{ demo }}</p>
        <p  :class="cnt">{{ jojo }}</p>
        
    </div>


</template>

<style scoped>

    .rd {
        font-size: 30px;
        color: blueviolet;
    }

</style>

<script>

export default {
    data() {
        return {
            
            }
    },
    
    props:{
        title:{
            type:[String,Number,Object,Array],

            // 不传的时候终端就会报警告，必选项
            required:true
        },

        // 没传的时候就设置默认值
        age:{
            type:Number,
            default:0
        },

        // 数组和对象默认值采用函数的形式
        name:{
            type:Array,
            default(){
                return ["空的"]
            }
        }
    }

}

</script>

<template>

    <div>
        <h2>B</h2>

        <!-- 父元素传过来的数据不可修改 -->
        <p>{{ title }}</p>
        <p>{{ age }}</p>
        <ul>
            <li v-for="(value, index) in name" :key="index">{{ value }} -- {{ index }}</li>
        </ul>
    </div>


</template>

### 组件事件

``` javascript
<script>
import index_4_Event from './index_4_Event.vue';

export default {
    data() {
        return {
            cnt:"content",
            
            
            message:"",
        }
    },

    components:{
        index_4_Event
    },
    methods:{
        getEvent(data){
            this.message = data;
        }
    }
}

</script>

<template>

    <div :class="cnt">
        <h2>组件事件</h2>
        <!-- 可以实现子传父 -->
        
        <index_4_Event @someEvent="getEvent"/>
        <p>父元素：{{ message }}</p>
    </div>


</template>

<style scoped>
    .content {
        display: flex;
        flex-direction: column;
        align-items: center;   
        justify-content: center;
        
        height: 50vh; 
        border: 1px solid #000;
    }
</style>

``` javascript
<script>
export default {
    data() {
        return {
            msg: "child component"
        }
    },

    methods:{
        clikc_SendEvent(){
            this.$emit("someEvent", this.msg)
        }
    }
}

</script>

<template>

    <div>
        <h2>自定义事件</h2>
        <button @click="clikc_SendEvent">发送数据到父元素</button>
        
    </div>


</template>

### 组件事件配合v-model使用

``` javascript
<script>
import index_5_Search from './index_5_Search.vue';

export default {
    data() {
        return {
            cnt:"content",

            val:"",
        }
    },

    components:{
        index_5_Search
    },
    methods: {
        getSearch(data){
            this.val = data;
        }
    }

}

</script>

<template>

    <div :class="cnt">
        <h2>组件事件配合v-model使用</h2>
        <index_5_Search @searchEvent="getSearch"/>
        <h3>搜索子组件获取到的内容：{{ val }}</h3>
    </div>


</template>

<style scoped>
    .content {
        display: flex;
        flex-direction: column;
        align-items: center;   
        justify-content: center;
        
        height: 50vh; 
        border: 1px solid #000;
    }
</style>

``` javascript
<script>

export default {
    data() {
        return {
            search: '',
        }
    },
    watch:{
        search(newVal){
            this.$emit('searchEvent', newVal)
        }
    }
    

}

</script>

<template>
    <h2>Main</h2>
    <h3>搜索框</h3>
    <input type="text" v-model="search">
</template>

### 使用Props方法子传父

``` javascript
<script>
import index_6_ComB from './index_6_ComB.vue';
    export default {
        data() {
            return {
                cnt:"content",
                msg:"",
            }
        },
        components: {
            index_6_ComB
        },
        methods: {
            onEvent(msg) {
                this.msg = msg
            }
        }
    }
</script>
<template>
    <div :class="cnt">
        <h3>使用Props方法子传父</h3>
        <index_6_ComB :sendMsg="onEvent" title="JOJO"/>
        <p>子组件传来的数据：{{msg}}</p>
    </div>
</template>

<style scoped>
    .content {
        display: flex;
        flex-direction: column;
        align-items: center;   
        justify-content: center;
        
        height: 50vh; 
        border: 1px solid #000;
    }
</style>

``` javascript
<script>
    export default {
        data() {
            return {
                msg:"send the data",
            }
        },
        props:{
            title:{
                type:String,
            },
            sendMsg:{
                type:Function,
            }
        }
    }
</script>
<template>
    <h3>子组件</h3>
    <p>{{ sendMsg(msg) }}</p>
</template>
