### 模板引用

``` javascript
<script>
export default {
    data() {
        return {
           cnt: "content",

            contents: "HI JOJO",

        }
    },

    methods: {
       getDemo(){
        console.log(this.$refs.cont.innerHTML = "Hello JOJO")
        console.log(this.$refs.username.value)
       }

    }
}

</script>

<template>
    <div :class="cnt">
        <h2>模板引用</h2>

        <div ref="cont" class="cont">
            {{ contents }}
        </div>
        

        <input type="text" ref="username">

        <button @click="getDemo">获取元素</button>
        
    </div>


</template>

<style>

    .content {
        display: flex;
        flex-direction: column;
        align-items: center;   
        justify-content: center;
        
        height: 50vh; 
        border: 1px solid #000;
    }

</style>

### 组件注册方式

> 分为全局注册和局部注册，推荐使用局部，全局无论后期是否使用都会一起打包

### 组件传递数据

> 需要注意的是 **只能父传子不能子传父**

> 先是父文件

``` javascript
<script>
import Child from "./index_2_Child.vue"

export default {
    data() {
        return {
            cnt:"content",

            test:"动态数据",
        }
    },

    components:{
        Child
    }
}

</script>

<template>

    <div :class="cnt">
        <h2>组件传递数据-Parent</h2>

        <child title="Parent 数据传递" demo="测试" :jojo = "test" />
        
    </div>


</template>

<style scoped>
    .content {
        display: flex;
        flex-direction: column;
        align-items: center;   
        justify-content: center;
        
        height: 50vh; 
        border: 1px solid #000;
    }
</style>

> 然后是子程序

``` javascript
<script>
export default {
    data() {
        return {
            cnt:"rd",
        }
    },

    props:["title","demo","jojo"],
}

</script>

<template>
    <div>
        <h2>组件传递数据-Child</h2>

        <p>{{ title }}</p>
        <p>{{ demo }}</p>
        <p  :class="cnt">{{ jojo }}</p>
        
    </div>


</template>

<style scoped>

    .rd {
        font-size: 30px;
        color: blueviolet;
    }

</style>

``` javascript
<script>

export default {
    data() {
        return {
            
            }
    },
    
    props:{
        title:{
            type:[String,Number,Object,Array],

            // 不传的时候终端就会报警告，必选项
            required:true
        },

        // 没传的时候就设置默认值
        age:{
            type:Number,
            default:0
        },

        // 数组和对象默认值采用函数的形式
        name:{
            type:Array,
            default(){
                return ["空的"]
            }
        }
    }

}

</script>

<template>

    <div>
        <h2>B</h2>

        <!-- 父元素传过来的数据不可修改 -->
        <p>{{ title }}</p>
        <p>{{ age }}</p>
        <ul>
            <li v-for="(value, index) in name" :key="index">{{ value }} -- {{ index }}</li>
        </ul>
    </div>


</template>
