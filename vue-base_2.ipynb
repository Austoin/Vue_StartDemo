``` javascript
<script>

    data() {
        return {
           cnt: "content",

           colorDemo: "red",
           fontSize: "30px",

           styleBoth:{
             color: "red",
             fontSize: "30px",
           }

        }
    },

    methods: {
       

    },

    computed: {


    }
}

</script>

<template>
    <div :class="cnt">
        <h2>Style 绑定</h2>

        <p :style="{color: colorDemo, fontSize: fontSize}">JOJO</p>

        <!-- 绑定对象 -->
        <p :style="styleBoth">JOJO</p>

        <!-- 绑定数组 -->
