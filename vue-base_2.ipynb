### style绑定

``` javascript
<script>
export default {
    data() {
        return {
           cnt: "content",

           colorDemo: "red",
           fontSize: "30px",

           styleBoth:{
             color: "red",
             fontSize: "30px",
           }

        }
    },

    methods: {
       

    },

    computed: {


    }
}

</script>

<template>
    <div :class="cnt">
        <h2>Style 绑定</h2>

        <p :style="{color: colorDemo, fontSize: fontSize}">JOJO</p>

        <!-- 绑定对象 -->
        <p :style="styleBoth">JOJO</p>

        <!-- 绑定数组 -->
        <p :style="[styleBoth, ]">HI</p>

        <!-- 但更加推荐使用class，style权限太高了后期不易更改 -->
        
    </div>


</template>

<style>

    .content {
        display: flex;
        flex-direction: column;
        align-items: center;   
        justify-content: center;
        
        height: 50vh; 
        border: 1px solid #000;
    }

</style>

### 侦听器

``` javascript
<script>
export default {
    data() {
        return {
           cnt: "content",

           message: "Hello JOJO",

        }
    },

    methods: {
       
        ChangeMethods() {
            this.message = "Hi JOJO"
        }

    },

    watch: {
        message(newData, oldData){
            console.log(newData, oldData)
        }
    }
}

</script>

<template>
    <div :class="cnt">
        <h2>侦听器</h2>

        <p>{{message}}</p>
        <button @click="ChangeMethods">change</button>
        
    </div>


</template>

<style>

    .content {
        display: flex;
        flex-direction: column;
        align-items: center;   
        justify-content: center;
        
        height: 50vh;

    }

</style>

### 表单输入绑定

``` javascript
<script>
export default {
    data() {
        return {
           cnt: "content",

           message: "",
           checked: false,

        }
    },

    methods: {
       

    }
}

</script>

<template>
    <div :class="cnt">
        <h2>表单输入绑定</h2>

        <form>

            <input type="text" v-model.lazy="message">  <!-- 回车后显示p标签 -->
            <p>{{ message }}</p>

            <input type="checkbox", id="checkbox" v-model="checked">
            <label for="checkbox">{{ checked }}</label>
        
        
        </form>
        
    </div>


</template>

<style>

    .content {
        display: flex;
        flex-direction: column;
        align-items: center;
